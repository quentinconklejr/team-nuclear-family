```python
import pandas as pd
from pathlib import Path
```

```python
processed_dir = Path.cwd().parent / "processed_data"
df = pd.read_csv(processed_dir / "final_dataset.csv", index_col=0)
print(f"Total counties: {len(df)}")
df.head()
```

```python
# drop counties with high seismic risk
seismic_ok = df["pga_max"] <= 0.3
df = df[seismic_ok]
print(f"{len(df)} counties remaining")
```

```python
# drop counties where too much land is a flood zone
flood_ok = df["pct_sfha"] <= 0.2
df = df[flood_ok]
print(f"{len(df)} counties remaining")
```

```python
# drop counties where too much land is military
military_ok = df["pct_military"] <= 0.1
df = df[military_ok]
print(f"{len(df)} counties remaining")
```

```python
# drop counties where too much land is protected (GAP status 1 or 2)
protected_ok = df["pct_protected"] <= 0.1
df = df[protected_ok]
print(f"{len(df)} counties remaining")
```

```python
df = df.reset_index(drop=True)
df
```

```python
df.to_csv(processed_dir / "candidates.csv", index=False)
```
