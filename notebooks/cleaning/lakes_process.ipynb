```python
# imports
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import geopandas as gpd
```

```python
lakes_data = Path.cwd().parent.joinpath('raw_data', 'lakes', 'lakes_us.shp')
lakes_data
```

```python
county_data = Path.cwd().parent.joinpath('raw_data', 'county_boundaries_2025', 'tl_2025_us_county.shp')
county_data
```

```python
df = gpd.read_file(lakes_data)
df.head()
```

```python
print(df.shape)
print(df.columns)
print(df.isnull().sum())
```

```python
df["Lake_type"].unique()
```

```python
df.crs
```

```python
df = df[["Hylak_id", "Lake_name", "Lake_area", "Vol_total", "Depth_avg", "Dis_avg", "geometry"]].to_crs(epsg = 5070)
df.columns = ["lake_id", "name", "area", "volume", "avg_depth", "avg_discharge", "geometry"]
df.head()
```

```python
assert df["lake_id"].nunique() == len(df)
```

```python
df_county = gpd.read_file(county_data)
print(df_county.shape)
df_county = df_county[["GEOID", "NAMELSAD", "geometry"]].rename(columns = {"GEOID": "geoid", "NAMELSAD": "county_name"})
df_county = df_county.to_crs(epsg = 5070)
df_county
```

```python
df_join = gpd.sjoin(df, df_county, predicate = "intersects", how = "left")
df_join.head()
```

```python
joined = gpd.sjoin(df_county[["geoid", "county_name", "geometry"]], 
                   df, how="left", predicate="intersects")

lake_features = joined.groupby("geoid").agg(
    lake_count      = ("lake_id", "count"),
    total_lake_area = ("area", "sum"),
    avg_vol         = ("volume", "mean"),
    avg_depth       = ("avg_depth", "mean"),
    avg_discharge   = ("avg_discharge", "mean")
).reset_index()

# Merge back to county
df_county = df_county.merge(lake_features, on="geoid", how="left")
df_county[["lake_count", "total_lake_area", "avg_vol", "avg_depth", "avg_discharge"]] = \
    df_county[["lake_count", "total_lake_area", "avg_vol", "avg_depth", "avg_discharge"]].fillna(0)

# county centroids
df_county["centroid"] = df_county.centroid
```

```python
df_county
```

```python
lakes_combined = df.geometry.union_all()
df_county["dist_to_lakes_km"] = df_county["centroid"].apply(
    lambda x: x.distance(lakes_combined) / 1000
)
```
