```python
high_risk_with_plants = df[
    (df['pga_max'] >= 0.3) & 
    (df['plant_count'] > 0)
]

high_risk_with_plants
```

```python
df_plot = df.copy()

def categorize(row):
    if row['pga_max'] >= 0.3 and row['plant_count'] > 0:
        return 'High Risk + Plant'
    elif row['pga_max'] >= 0.3:
        return 'High Risk Only'
    elif row['plant_count'] > 0:
        return 'Plant Only'
    else:
        return 'Neither'


df_plot['category'] = df_plot.apply(categorize, axis=1)

sns.relplot(
    data=df_plot,
    x='total_energy_consumption_mwh',
    y='pga_max',
    col='category',
    hue='category',
    col_wrap=2,
    kind='scatter',
    alpha=0.6
).set(xscale="log")
```

#### Seismic Risk, Energy Demand, and Presence of Plant (Log Scale)

This visualization compares seismic risk (pga_max) and energy demand (total_energy_consumption_mwh) across four categories:

Neither: low risk, no plant
High Risk Only: high risk, no plant
Plant Only: low risk, has plant
High Risk + Plant: high risk, has plant

The x-axis is plotted on a log scale to account for the wide range of energy consumption values. Without transformation, counties with extremely large consumption would compress the majority of observations near zero, making patterns difficult to observe.

```python
Q1_distance_to_rivers_km = df['distance_to_rivers_km'].quantile(0.25)
Q3_distance_to_rivers_km = df['distance_to_rivers_km'].quantile(0.75)
IQR_distance_to_rivers_km = Q3_distance_to_rivers_km - Q1_distance_to_rivers_km

distance_to_river_non_outliers = df['distance_to_rivers_km'].between(
    Q1_distance_to_rivers_km - 1.5 * IQR_distance_to_rivers_km,
    Q3_distance_to_rivers_km + 1.5 * IQR_distance_to_rivers_km
)

import seaborn as sns

sns.boxplot(y=df.loc[distance_to_river_non_outliers, 'distance_to_rivers_km'])
```
